```python
import tensorflow as tf
from tensorflow import keras
from keras.layers import Dense, Conv2D, Flatten, Input
from keras import Sequential
from keras.datasets import mnist
```

```python
# Load dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
# 1. Model using VALID padding (No padding)
# Input shape: (28, 28, 1)
model_valid = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, kernel_size=(3, 3), padding='valid', activation='relu'),
    Conv2D(32, kernel_size=(3, 3), padding='valid', activation='relu'),
    Conv2D(32, kernel_size=(3, 3), padding='valid', activation='relu'),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

print(" VALID PADDING MODEL SUMMARY ")
model_valid.summary()
```

```python
# 2. Model using SAME padding (Zero padding added)
# Input shape: (28, 28, 1)
model_same = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, kernel_size=(3, 3), padding='same', activation='relu'),
    Conv2D(32, kernel_size=(3, 3), padding='same', activation='relu'),
    Conv2D(32, kernel_size=(3, 3), padding='same', activation='relu'),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

print("\n SAME PADDING MODEL SUMMARY ")
model_same.summary()
```

```python
# 1. Model using VALID Padding and Stride = 1 (Default behavior)
model_valid_s1 = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, kernel_size=(3, 3), strides=(1, 1), padding='valid', activation='relu'),
    Conv2D(32, kernel_size=(3, 3), strides=(1, 1), padding='valid', activation='relu'),
    Flatten(),
    Dense(10, activation='softmax')
])

print("=== VALID PADDING (Stride=1) SUMMARY ===")
model_valid_s1.summary()
```

```python
# 2. Model using VALID Padding and Stride = 2 (Downsampling without padding)
model_valid_s2 = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, kernel_size=(3, 3), strides=(2, 2), padding='valid', activation='relu'),
    Conv2D(32, kernel_size=(3, 3), strides=(2, 2), padding='valid', activation='relu'),
    Flatten(),
    Dense(10, activation='softmax')
])

print("\n=== VALID PADDING (Stride=2) SUMMARY ===")
model_valid_s2.summary()
```

```python
# 3. Model using SAME Padding and Stride = 2 (Downsampling with padding)
model_same_s2 = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, kernel_size=(3, 3), strides=(2, 2), padding='same', activation='relu'),
    Conv2D(32, kernel_size=(3, 3), strides=(2, 2), padding='same', activation='relu'),
    Flatten(),
    Dense(10, activation='softmax')
])

print("\n=== SAME PADDING (Stride=2) SUMMARY ===")
model_same_s2.summary()
```
